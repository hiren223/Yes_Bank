Overall, this project combines data preprocessing, exploratory analysis, feature engineering, regression modelling, and model evaluation to build a structured machine learning solution for Yes Bank monthly stock closing price prediction.

## ***5. Hypothesis Testing***

### Based on your chart experiments, define three hypothetical statements from the dataset. In the next three questions, perform hypothesis testing to obtain final conclusion about the statements through your code and statistical testing.

Hypothesis Statements – Yes Bank Stock Price Prediction
Based on the charts and the 185 monthly observations in your dataset, the following three hypotheses can be tested statistically.

We will use a 5% significance level (α = 0.05) for all three tests.

### Hypothetical Statement - 1

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

Relationship between Opening Price and Closing Price

Statement: The monthly opening price has a significant positive relationship with the monthly closing price of Yes Bank.
- Null Hypothesis (H₀): There is no significant correlation between Opening Price and Closing Price.
- Alternative Hypothesis (H₁): There is a significant positive correlation between Opening Price and Closing Price.
- Statistical test: Pearson correlation test.
- Chart reference: Opening Price vs Closing Price (Scatter Plot).

#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value


from scipy.stats import pearsonr

# Define significance level
alpha = 0.05

# Extract variables
x = df['Open']
y = df['Close']

# Perform one-tailed Pearson correlation test
correlation, p_value = pearsonr(x, y, alternative='greater')

print("Pearson Correlation Test")
print("--------------------------------")
print("Correlation Coefficient:", correlation)
print("P-Value:", p_value)
print("Significance Level:", alpha)

if p_value < alpha:
    print("Reject the Null Hypothesis (H0)")
    print("There is a significant positive correlation.")
else:
    print("Fail to Reject the Null Hypothesis (H0)")
    print("There is insufficient evidence of a significant positive correlation.")
````

##### Which statistical test have you done to obtain P-Value?

I performed the Pearson Correlation Test to measure the strength and direction of the linear relationship between the monthly Opening Price and Closing Price of Yes Bank.
Since the alternative hypothesis specifically predicts a positive relationship, I used a one-tailed Pearson correlation test.

Hypotheses:
- H0(Null Hypothesis): There is no significant positive correlation between Opening Price and Closing Price.
- H1(Alternative Hypothesis): There is a significant positive correlation between Opening Price and Closing Price.

##### Why did you choose the specific statistical test?

The Pearson Correlation Test was selected because:
1. Both Opening Price and Closing Price are continuous numerical variables.
2. The objective is to measure the strength and direction of their linear relationship.
3. The scatter plot can help identify whether the relationship is approximately linear.
4. The test provides a correlation coefficient (r) and p-value to determine whether the observed positive relationship is statistically significant.

### Conclusion – Hypothesis 1

The Pearson Correlation Test was performed to examine the relationship between the monthly Opening Price and Closing Price of Yes Bank.

The calculated correlation coefficient is 0.9780, indicating a very strong positive linear relationship. The p-value is approximately 9.26 × 10⁻¹²⁷, which is significantly lower than the significance level of 0.05.

Therefore, we reject the null hypothesis (H₀) and accept the alternative hypothesis (H₁).

**Final Result:** There is strong statistical evidence of a significant positive correlation between the monthly Opening Price and Closing Price of Yes Bank.

````python
# Perform Statistical Test to obtain P-Value

import pandas as pd
import numpy as np
from scipy.stats import ttest_ind

# Convert Date into datetime format
df['Date'] = pd.to_datetime(df['Date'], format='%b-%y')

# Sort data chronologically
df = df.sort_values('Date').reset_index(drop=True)

# Calculate monthly percentage returns
df['Monthly_Return'] = df['Close'].pct_change() * 100

# Divide data into two periods
before_2018 = df.loc[
    df['Date'] < '2018-07-01',
    'Monthly_Return'
].dropna()

after_2018 = df.loc[
    df['Date'] >= '2018-07-01',
    'Monthly_Return'
].dropna()

# Perform independent two-sample Welch's t-test
t_statistic, p_value = ttest_ind(
    before_2018,
    after_2018,
    equal_var=False,
    alternative='two-sided'
)

# Display results
print("Welch's Independent Two-Sample T-Test")
print("--------------------------------------")

print("Before July 2018:")
print("Observations:", len(before_2018))
print("Mean Monthly Return:", before_2018.mean())
print("Standard Deviation:", before_2018.std())

print("\nJuly 2018 Onward:")
print("Observations:", len(after_2018))
print("Mean Monthly Return:", after_2018.mean())
print("Standard Deviation:", after_2018.std())

print("\nT-Statistic:", t_statistic)
print("P-Value:", p_value)

alpha = 0.05
print("Significance Level:", alpha)

if p_value < alpha:
    print("\nReject the Null Hypothesis (H0)")
    print("There is a significant difference in average monthly returns.")
else:
    print("\nFail to Reject the Null Hypothesis (H0)")
    print("There is insufficient evidence of a significant difference.")
````

##### Which statistical test have you done to obtain P-Value?

I performed an Independent Two-Sample Welch's t-test to determine whether the average monthly stock returns of Yes Bank differ between the two periods:
- Before July 2018: Monthly returns dated before July 2018.
- July 2018 onward: Monthly returns dated July 2018 or later.

Welch's t-test compares the means of two independent groups without assuming that their variances are equal.

**Hypotheses:**
- H₀ (Null Hypothesis): There is no significant difference in average monthly returns between the two periods.
- H₁ (Alternative Hypothesis): There is a significant difference in average monthly returns between the two periods.

##### Why did you choose the specific statistical test?

Welch's independent two-sample t-test was selected because:
1. Monthly returns are continuous numerical variables.
2. The data is divided into two separate time periods.
3. The objective is to compare the average monthly returns of these two groups.
4. Welch's t-test does not require equal variances between the groups, making it more suitable when volatility differs between periods.

### Conclusion – Hypothesis 3

Welch's Independent Two-Sample t-test was performed to compare the average monthly stock returns of Yes Bank before July 2018 and from July 2018 onward.

The average monthly return before July 2018 was approximately 3.06%, while the average monthly return from July 2018 onward was approximately -6.88%.

The calculated t-statistic was 2.0340, and the p-value was 0.05056. Since the p-value is slightly greater than the significance level of 0.05, we fail to reject the null hypothesis (H₀).

**Final Result:** At the 5% significance level, there is insufficient statistical evidence to conclude that the average monthly returns differed significantly between the two periods. However, the observed average returns show a substantial negative change after July 2018, which deserves further investigation.

#### 1. Feature Manipulation

````python
# Manipulate Features to minimize feature correlation and create new features

# Manipulate Features to minimize feature correlation
# and create new features

# Create a separate modelling dataset
df_model = df[
    ['Date', 'Open', 'High', 'Low', 'Close']
].copy()

# Convert Date into datetime
df_model['Date'] = pd.to_datetime(
    df_model['Date'],
    format='%b-%y'
)

# Sort chronologically
df_model = df_model.sort_values(
    'Date'
).reset_index(drop=True)

# 1. Monthly percentage return
df_model['Monthly_Return'] = (
    df_model['Close'].pct_change() * 100
)

# 2. Monthly price range
df_model['Price_Range'] = (
    df_model['High'] - df_model['Low']
)

# 3. Opening and closing price difference
df_model['Open_Close_Diff'] = (
    df_model['Close'] - df_model['Open']
)

# 4. Previous month's closing price
df_model['Close_Lag1'] = (
    df_model['Close'].shift(1)
)

# 5. Closing price two months earlier
df_model['Close_Lag2'] = (
    df_model['Close'].shift(2)
)

# 6. Previous month's return
df_model['Return_Lag1'] = (
    df_model['Monthly_Return'].shift(1)
)

# 7. Three-month rolling average
df_model['Rolling_Mean_3'] = (
    df_model['Close'].rolling(window=3).mean()
)

# 8. Three-month rolling standard deviation
df_model['Rolling_Std_3'] = (
    df_model['Close'].rolling(window=3).std()
)

# Target: Next month's closing price
df_model['Target_Close'] = (
    df_model['Close'].shift(-1)
)

display(df_model.head(10))
````

````python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Ridge
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor

from sklearn.model_selection import GridSearchCV, TimeSeriesSplit

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from sklearn.inspection import permutation_importance

import warnings
warnings.filterwarnings('ignore')
````

````python

print("Training Features:", X_train.shape)
print("Testing Features:", X_test.shape)

print("Training Target:", y_train.shape)
print("Testing Target:", y_test.shape)

print("\nTraining date range:")
print(
    df_model_clean['Date'].iloc[0],
    "to",
    df_model_clean['Date'].iloc[len(X_train)-1]
)

print("\nTesting date range:")
print(
    df_model_clean['Date'].iloc[len(X_train)],
    "to",
    df_model_clean['Date'].iloc[-1]
)
````

````python

def evaluate_model(y_true, y_pred, model_name):

    mae = mean_absolute_error(y_true, y_pred)
    mse = mean_squared_error(y_true, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_true, y_pred)

    return {
        'Model': model_name,
        'MAE': mae,
        'MSE': mse,
        'RMSE': rmse,
        'R2 Score': r2
    }
````

````python
# ML Model - 1 Implementation
ridge_model = Ridge(alpha=1.0)

# Fit the Algorithm
ridge_model.fit(X_train_scaled, y_train)

# Predict on the model
ridge_predictions = ridge_model.predict(X_test_scaled)

#Evalute model
ridge_results = evaluate_model(
    y_test,
    ridge_predictions,
    'Ridge Regression'
)

print(ridge_results)
````

Ridge Regression is a linear regression algorithm that uses L2 regularization to reduce the effect of multicollinearity and overfitting.
